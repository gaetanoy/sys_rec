```python
# Importer les bibliothèques nécessaires
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, accuracy_score, classification_report
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# Création d'un petit dataset pour les exercices
data = {
    'Age': [22, 25, 47, 52, 46, 56, 23, 40, 33, 55],
    'Income': [15000, 29000, 54000, 63000, 58000, 76000, 19000, 52000, 33000, 72000],
    'Education_Level': [0, 1, 2, 2, 1, 2, 0, 2, 1, 2],  # 0: High School, 1: Bachelor, 2: Graduate
    'Purchase': [0, 0, 1, 1, 1, 1, 0, 1, 0, 1]  # 0: No, 1: Yes
}

# Convertir en DataFrame
df = pd.DataFrame(data)

# Séparer les caractéristiques et la variable cible
X = df[['Age', 'Income', 'Education_Level']]
y = df['Purchase']

# Diviser le dataset en ensemble d'entraînement et de test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
df.head()
```

## A. Apprentissage Supervisé 
## 1. Régression Linéaire
Utilisez la régression linéaire pour prédire une variable continue (par exemple, le revenu 'Income').


```python
# Initialiser et entraîner le modèle
linear_reg = LinearRegression()
linear_reg.fit(X_train[['Age', 'Education_Level']], X_train['Income'])

# Prédictions
income_pred = linear_reg.predict(X_test[['Age', 'Education_Level']])

# Évaluer le modèle
mse = mean_squared_error(X_test['Income'], income_pred)
print("Mean Squared Error:", mse)
```

## 2. Régression Logistique
Utilisez la régression logistique pour prédire si un client achètera un produit ou non.


```python
# Initialiser et entraîner le modèle
logistic_reg = LogisticRegression()
logistic_reg.fit(X_train, y_train)

# Prédictions
y_pred = logistic_reg.predict(X_test)

# Évaluer le modèle
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

## 3. k-Nearest Neighbors (k-NN)
Utilisez k-NN pour classifier si un client achètera un produit.


```python
# Initialiser le modèle avec k=3
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)

# Prédictions
y_pred = knn.predict(X_test)

# Évaluation
print(classification_report(y_test, y_pred))
```

## 4. Support Vector Machines (SVM)
Utilisez SVM pour classifier si un client achètera un produit.


```python
# Initialiser le modèle avec un noyau RBF
svm = SVC(kernel='rbf', gamma='auto')
svm.fit(X_train, y_train)

# Prédictions
y_pred = svm.predict(X_test)

# Évaluation
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

## 5. Arbres de Décision
Utilisez un arbre de décision pour classifier si un client achètera un produit.


```python
# Initialiser le modèle
decision_tree = DecisionTreeClassifier(max_depth=3)
decision_tree.fit(X_train, y_train)

# Prédictions
y_pred = decision_tree.predict(X_test)

# Évaluation
print(classification_report(y_test, y_pred))
```

## 6. Forêts Aléatoires
Utilisez des forêts aléatoires pour classifier si un client achètera un produit.


```python
# Initialiser le modèle
random_forest = RandomForestClassifier(n_estimators=100, max_depth=3, random_state=42)
random_forest.fit(X_train, y_train)

# Prédictions
y_pred = random_forest.predict(X_test)

# Évaluation
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

## Créer un petit dataset et appliquer K-Means avec 2 clusters

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

# Creating a sample dataset
X = np.array([[1, 2], [2, 3], [3, 4], [8, 8], [9, 10], [10, 12]])

# Applying K-Means with 2 clusters
kmeans = KMeans(n_clusters=2, random_state=0).fit(X)
labels = kmeans.labels_
centroids = kmeans.cluster_centers_

# Plotting the clusters and centroids
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='viridis', marker='o')
plt.scatter(centroids[:, 0], centroids[:, 1], c='red', marker='x', s=100)
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('K-Means Clustering')
plt.show()
```

```python
from sklearn.datasets import make_blobs

# Générer un jeu de données artificiel
X, y = make_blobs(n_samples=300, centers=4, cluster_std=2.0, random_state=42)

# Visualiser les données
plt.scatter(X[:, 0], X[:, 1], s=30, cmap='viridis')
plt.title("Données initiales")
plt.show()

# Calculer l'inertie pour différents nombres de clusters
inertias = []
K_range = range(1, 11)

for k in K_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X)
    inertias.append(kmeans.inertia_)

# Tracer l'inertie en fonction du nombre de clusters
plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, marker='o', linestyle='--', color='b')
plt.title('Méthode du coude pour choisir K')
plt.xlabel('Nombre de clusters (K)')
plt.ylabel('Inertie')
plt.xticks(K_range)
plt.grid()
plt.show()
```

### b) Hierarchical Clustering

```python
from scipy.cluster.hierarchy import dendrogram, linkage
import matplotlib.pyplot as plt

# Generating a sample dataset
X = np.array([[1, 2], [2, 3], [3, 4], [8, 8], [9, 10], [10, 12]])

# Performing hierarchical clustering
linked = linkage(X, method='ward')  # 'ward' minimizes the variance within clusters

# Plotting the dendrogram
plt.figure(figsize=(8, 4))
dendrogram(linked, orientation='top', distance_sort='descending', show_leaf_counts=True)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Sample index')
plt.ylabel('Distance')
plt.show()
```

### 2. Les techniques de réduction de dimension
###  a) PCA

```python
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Sample data with high dimensionality
X = np.array([[1, 2, 3], [2, 3, 5], [3, 4, 2], [4, 5, 8], [5, 6, 5], [6, 7, 8]])

# Applying PCA to reduce to 2 dimensions
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

# Plotting the reduced data
plt.scatter(X_pca[:, 0], X_pca[:, 1], c='blue', marker='o')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA - Dimensionality Reduction')
plt.show()
```
